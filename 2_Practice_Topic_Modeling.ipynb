# Exercise 1

Latent Semantic Indexing (LSI) models underlying concepts by using SVD (Singular Value Decomposition).

Use [gensim](https://radimrehurek.com/gensim/) library to:
1. Create a corpus composed of the headlines contained in the data collection.
2. Generate a [dictionary](https://radimrehurek.com/gensim/corpora/dictionary.html) to create a word -> id mapping (required by LSI module).
3. Using the dictionary, preprocess the corpus to obtain the representation required for LSI model training ([documentation here](https://radimrehurek.com/gensim/models/lsimodel.html)).
4. Inspect the top-5 topics generated by the LSI model for the analysed corpus.

````python
#! pip uninstall numpy
#! pip install numpy==1.26.4
#! pip install --upgrade scipy gensim
````

````python
import pandas as pd
df_tmodeling = pd.read_csv('CovidFake_filtered.csv')
````

````python
# generating the corpus
corpus = df_tmodeling["headlines"].tolist()
corpus = [s.split() for s in corpus]
print (corpus[:10])
````

````python
# constructing dictionary and processing the corpus
from gensim.corpora.dictionary import Dictionary
tm_dict = Dictionary(corpus)
processed_corpus = [tm_dict.doc2bow(text) for text in corpus]
````

````python
# training LSI model
from gensim.models import LsiModel
model = LsiModel(processed_corpus, id2word=tm_dict)
model.print_topics(5)
````

# Exercise 2

Leveraging the same corpus used for LSI model generation, apply LDA modeling setting the number of topics to 5. Display the words most contributing to the those topics according to the LDA model.

````python
from gensim.models.ldamodel import LdaModel
lda = LdaModel(processed_corpus, id2word=tm_dict, num_topics=5)
lda.print_topics(5)
````

# Exercise 3

The top-scored words contributing to each topic (if no stopword removal is applied) are english common words (e.g., *to, for, in, of, on*..). Moreover, missing punctuation removal could be critical for topic identification. Add preliminary preprocessing step to:
1. **remove stopwords**
2. **strip punctuation**
3. **lowercase all words**

**Suggestion**: many predefined lists of stopwords are already available. Download the stopword list of NLTK directly in your python code.

````python
## YOUR CODE HERE

import nltk
import string
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
def preprocess(doc):
    tokens = doc.lower().split()
    tokens = [t.translate(str.maketrans('', '', string.punctuation)) for t in tokens]
    tokens = [t for t in tokens if t and t not in stop_words]
    return tokens

rs_corpus = [preprocess(doc) for doc in df_tmodeling["headlines"].tolist()]

# Rebuild dictionary and processed corpus with cleaned text
tm_dict = Dictionary(rs_corpus)
processed_corpus = [tm_dict.doc2bow(text) for text in rs_corpus]

# Re-train LSI and LDA on cleaned corpus
model = LsiModel(processed_corpus, id2word=tm_dict)
print("LSI topics after preprocessing:")
model.print_topics(5)

lda = LdaModel(processed_corpus, id2word=tm_dict, num_topics=5)
print("\nLDA topics after preprocessing:")
lda.print_topics(5)
````

Are the documents grouped correctly? Can you think about a way to assign a label to each cluster?

**Suggestions:**
1.   Create the embedding of each document
2.   Pass the embedding to clustering algorithm
3.   Manually check cluster quality




````python
## YOUR CODE HERE
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import numpy as np

docs = [
    "Apple unveils new AI features for iPhone users",
    "Microsoft launches a new cloud security platform",
    "Stock markets fall as inflation fears return",
    "Investors expect another interest rate increase",
    "Real Madrid wins after a dramatic extra-time goal",
    "Barcelona signs a new striker for next season",
    "Scientists discover a new method for battery recycling",
    "Researchers develop a model to detect fake news",
    "Parliament debates a new immigration law",
    "Government approves an education reform package",
    "NASA prepares its next mission to the Moon",
    "SpaceX tests a new reusable rocket system",
]

# Step 1: Embed documents using sentence-transformers (same model used by BERTopic above)
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
embeddings = model.encode(docs, show_progress_bar=True)

# Step 2: Cluster with KMeans (6 clusters = 6 natural topics)
kmeans = KMeans(n_clusters=6, random_state=42)
labels = kmeans.fit_predict(embeddings)

# Step 3: Display clusters with manual labels
cluster_names = {
    0: "Technology",
    1: "Finance",
    2: "Sports",
    3: "Science",
    4: "Politics",
    5: "Space",
}

clusters = {}
for doc, label in zip(docs, labels):
    clusters.setdefault(label, []).append(doc)

for cluster_id, cluster_docs in sorted(clusters.items()):
    print(f"\nCluster {cluster_id}:")
    for d in cluster_docs:
        print(f"  - {d}")
````
